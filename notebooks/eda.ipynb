# 信贷风控 EDA 探索性数据分析
1. 读取合成信贷数据
2. 查看基础信息、缺失值
3. 目标变量分布
4. 特征描述统计
5. 特征相关性分析


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
plt.rcParams['font.sans-serif'] = ['WenQuanYi Zen Hei']
plt.rcParams['axes.unicode_minus'] = False

df = pd.read_parquet('../data/credit_data.parquet')
print(df.shape)
df.head()
```

```python
# 缺失值检查
print(df.isnull().sum())
# 目标变量分布
print(df['target'].value_counts())
print(df['target'].value_counts(normalize=True))
```

```python
# 描述性统计
df.describe()
```

```python
# 相关性热力图
corr = df.corr()
plt.figure(figsize=(10,8))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('特征相关性热力图')
plt.show()
```

## EDA结论
- 数据集无缺失值
- 样本存在轻微不均衡，逾期样本占比偏高
- 负债比例、历史逾期是和target相关性最高的特征
